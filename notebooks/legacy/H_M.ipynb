```python
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from pmdarima import auto_arima
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from sklearn.metrics import mean_absolute_error, mean_squared_error
import os

# Create a new folder for saving results
output_folder = "Sales_Forecast_Results_with_Accuracy"
os.makedirs(output_folder, exist_ok=True)

# Load Dataset
file_path = "../../data/raw/hm.csv"  # Ensure this is the correct path
data = pd.read_csv(file_path, encoding="utf-8")

# Convert 'SaleDate' to datetime format
data['SaleDate'] = pd.to_datetime(data['SaleDate'], format='%d/%m/%Y')

# Get list of all unique products
products = data['ProductName'].unique()

# Function to calculate accuracy metrics
def calculate_accuracy_metrics(actual, predicted):
    mae = mean_absolute_error(actual, predicted)
    mse = mean_squared_error(actual, predicted)
    rmse = np.sqrt(mse)
    mape = np.mean(np.abs((actual - predicted) / actual)) * 100 if np.all(actual != 0) else np.nan
    return mae, mse, rmse, mape

# Function to create LSTM data
def create_lstm_data(series, time_steps=3):
    X, y = [], []
    for i in range(len(series) - time_steps):
        X.append(series[i:i + time_steps])
        y.append(series[i + time_steps])
    return np.array(X), np.array(y)

# Initialize a list to collect all metrics
all_metrics = []

# Loop through all products
for selected_product in products:
    print(f"Processing: {selected_product}...")

    # Extract and process data for the selected product
    product_data = data[data['ProductName'] == selected_product]

    # Aggregate sales by month
    monthly_sales = product_data.resample('M', on='SaleDate')['QuantitySold'].sum().fillna(0)

    # Ensure sufficient data for training
    if len(monthly_sales) < 24:
        print(f"Skipping {selected_product} due to insufficient data.")
        continue

    # Split data into training (2015-2022) and test (2023)
    train = monthly_sales["2015-01-31":"2022-12-31"]
    test = monthly_sales["2023-01-31":"2023-12-31"]

    if len(test) < 12:
        print(f"Skipping {selected_product} due to insufficient test data.")
        continue

    # Fit ARIMA model
    auto_model = auto_arima(train, seasonal=True, m=12, d=1, D=1, stepwise=True, trace=True)
    best_order = auto_model.order
    best_seasonal_order = auto_model.seasonal_order

    # Train ARIMA model
    arima_model = ARIMA(train, order=best_order, seasonal_order=best_seasonal_order)
    arima_results = arima_model.fit()

    # Forecast for the test period (12 months)
    test_forecast_steps = len(test)
    arima_forecast = arima_results.forecast(steps=test_forecast_steps)

    # Ensure train has enough data for residuals
    if len(train) < test_forecast_steps:
        print(f"Skipping {selected_product} due to insufficient training data for residuals.")
        continue

    # Calculate residuals
    residuals = train.values[-test_forecast_steps:] - arima_forecast[:test_forecast_steps].to_numpy()
    if len(residuals) < test_forecast_steps:
        print(f"Skipping {selected_product} due to insufficient residuals.")
        continue

    # Scale residuals
    scaler = StandardScaler()
    residuals_scaled = scaler.fit_transform(residuals.reshape(-1, 1))
    print(f"Shape of residuals_scaled: {residuals_scaled.shape}")

    # Create LSTM data
    time_steps = 3
    X_train, y_train = create_lstm_data(residuals_scaled, time_steps)
    print(f"Shape of X_train: {X_train.shape}")

    if len(X_train) == 0:
        print(f"Skipping {selected_product} due to insufficient LSTM training samples.")
        continue

    # Check available sequences for prediction
    available_sequences = X_train.shape[0]
    if available_sequences < test_forecast_steps:
        print(f"Warning: Only {available_sequences} sequences available for {selected_product}. Adjusting test_forecast_steps.")
        test_forecast_steps = min(test_forecast_steps, available_sequences)

    X_train = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))

    # Train LSTM model
    lstm_model = Sequential([
        LSTM(100, activation='relu', return_sequences=True, input_shape=(time_steps, 1)),
        Dropout(0.2),
        LSTM(50, activation='relu'),
        Dropout(0.2),
        Dense(1)
    ])
    lstm_model.compile(optimizer='adam', loss='mse')
    lstm_model.fit(X_train, y_train, epochs=100, batch_size=16, verbose=0)

    # Predict for test period
    X_pred = X_train[-test_forecast_steps:].reshape((test_forecast_steps, time_steps, 1))
    lstm_pred = lstm_model.predict(X_pred).flatten()
    lstm_pred = scaler.inverse_transform(lstm_pred.reshape(-1, 1)).flatten()
    final_forecast = 0.90 * arima_forecast[:test_forecast_steps].to_numpy() + 0.10 * lstm_pred
    final_forecast = np.maximum(1, np.round(final_forecast)).astype(int)

    # Calculate accuracy metrics
    actual_sales = test.values[:test_forecast_steps]
    mae, mse, rmse, mape = calculate_accuracy_metrics(actual_sales, final_forecast)

    print(f"Accuracy Metrics for {selected_product}:")
    print(f"MAE: {mae:.2f}")
    print(f"MSE: {mse:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"MAPE: {mape:.2f}%" if not np.isnan(mape) else "MAPE: Not computable (zero values in actual sales)")

    # Collect metrics for this product
    metrics_dict = {
        'Product': selected_product,
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'MAPE': mape if not np.isnan(mape) else 'Not computable'
    }
    all_metrics.append(metrics_dict)

    # Forecast for 2024-2028
    forecast_steps = 60
    arima_forecast_future = arima_results.forecast(steps=forecast_steps)
    residuals_future = arima_forecast_future.to_numpy()
    residuals_scaled_future = scaler.fit_transform(residuals_future.reshape(-1, 1))
    X_future, _ = create_lstm_data(residuals_scaled_future, time_steps)
    if X_future.shape[0] < forecast_steps:
        print(f"Warning: Only {X_future.shape[0]} sequences available for future forecast. Adjusting forecast_steps.")
        forecast_steps = min(forecast_steps, X_future.shape[0])

    X_future = X_future.reshape((X_future.shape[0], X_future.shape[1], 1))
    lstm_pred_future = lstm_model.predict(X_future).flatten()
    lstm_pred_future = scaler.inverse_transform(lstm_pred_future.reshape(-1, 1)).flatten()
    final_forecast_future = 0.90 * arima_forecast_future.to_numpy()[:len(lstm_pred_future)] + 0.10 * lstm_pred_future
    final_forecast_future = np.maximum(1, np.round(final_forecast_future)).astype(int)

    forecast_dates = pd.date_range(start=monthly_sales.index[-1], periods=len(final_forecast_future) + 1, freq='M')[1:]

    # Save forecast
    forecast_file = f"{output_folder}/{selected_product}_forecast_2024_2028.csv"
    pd.DataFrame({'Date': forecast_dates, 'Predicted Sales': final_forecast_future}).to_csv(forecast_file, index=False)
    print(f"Forecasting completed for {selected_product}!")
    print(f"Results saved in: {forecast_file}")

    # Plot actual and future forecast (remove test forecast)
    plt.figure(figsize=(14, 7))
    plt.plot(monthly_sales.index, monthly_sales.values, label="Actual Sales (2015-2023)", marker="o", color="blue")
    plt.plot(forecast_dates, final_forecast_future, label="Predicted Sales (2024-2028)", marker="o", linestyle="--", color="red")
    plt.plot([monthly_sales.index[-1], forecast_dates[0]], [monthly_sales.values[-1], final_forecast_future[0]], linestyle="--", color="red")
    plt.title(f"Sales Forecast for {selected_product}")
    plt.xlabel("Date")
    plt.ylabel("Quantity Sold")
    plt.legend()
    plt.grid(True)

    plot_file = f"{output_folder}/{selected_product}_forecast_plot.png"
    plt.savefig(plot_file, dpi=300, bbox_inches="tight")
    print(f"Forecast plot saved as: {plot_file}")
    plt.show()

# Save all metrics to a single CSV file
metrics_df = pd.DataFrame(all_metrics)
metrics_file = f"{output_folder}/H&M_accuracy_metrics.csv"
metrics_df.to_csv(metrics_file, index=False)
print(f"All accuracy metrics saved in: {metrics_file}")
```

```python
import pandas as pd
import os

# 📂 Define folders
actual_sales_file = "H_M.csv"  # Your original dataset
forecast_folder = "Sales_Forecast_Results_with_Accuracy"  # Folder containing forecasted sales
output_folder = "Sales_Forecasts_Merged_with_Accuracy"  # Folder to save merged results

# ✅ Create output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

# ✅ Load actual sales data
data = pd.read_csv(actual_sales_file)
data['SaleDate'] = pd.to_datetime(data['SaleDate'], format='%d/%m/%Y')

# ✅ Aggregate actual sales by month
actual_sales = data.groupby(['ProductName', pd.Grouper(key='SaleDate', freq='M')])['QuantitySold'].sum().reset_index()

# 🔹 Get list of all unique products
products = data['ProductName'].unique()

# 🔄 Loop through all products and merge actual & predicted sales
for selected_product in products:
    print(f"Processing: {selected_product}...")

    # ✅ Get actual sales for the product
    product_sales = actual_sales[actual_sales['ProductName'] == selected_product]

    # ✅ Try to load the forecasted sales CSV (Skip if not found)
    forecast_file = os.path.join(forecast_folder, f"{selected_product}_forecast_2024_2028.csv")
    if not os.path.exists(forecast_file):
        print(f"❌ No forecast data found for {selected_product}. Skipping.")
        continue
    
    forecast_data = pd.read_csv(forecast_file)
    forecast_data['Date'] = pd.to_datetime(forecast_data['Date'])

    # ✅ Merge actual and forecasted sales
    combined_sales = pd.concat([
        product_sales.rename(columns={'SaleDate': 'Date', 'QuantitySold': 'Sales'}),
        forecast_data.rename(columns={'Predicted Sales': 'Sales'})
    ])

    # ✅ Save merged sales data
    output_file = os.path.join(output_folder, f"{selected_product}_total_and_forecasted_sales.csv")
    combined_sales.to_csv(output_file, index=False)

    print(f"✅ Saved merged sales data for {selected_product}: {output_file}")
```

```python
import pandas as pd
import os

# 📂 Define the folder containing merged sales data
merged_folder = "Sales_Forecasts_Merged_with_Accuracy"  # Update if needed
output_file = "HM_All_Product_Sales.xlsx"  # The final Excel file

# ✅ Get list of all merged sales files
merged_files = [f for f in os.listdir(merged_folder) if f.endswith("_total_and_forecasted_sales.csv")]

# ✅ Read and combine all merged sales data
all_sales_data = []

for file in merged_files:
    product_name = file.replace("_total_and_forecasted_sales.csv", "")  # Extract product name
    file_path = os.path.join(merged_folder, file)
    
    # Load sales data
    df = pd.read_csv(file_path)
    df['Date'] = pd.to_datetime(df['Date'])
    df['ProductName'] = product_name  # Add product name column
    
    all_sales_data.append(df)

# ✅ Combine all products into a single DataFrame
merged_sales_df = pd.concat(all_sales_data, ignore_index=True)

# ✅ Save all product sales in **one Excel file**
merged_sales_df.to_excel(output_file, index=False, sheet_name="Sales_Data")

print(f"✅ All product sales saved in: {output_file}")

# ✅ Display first few rows to verify
print(merged_sales_df.head())
```
