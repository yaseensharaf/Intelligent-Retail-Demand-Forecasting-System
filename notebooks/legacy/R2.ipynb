```python
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from pmdarima import auto_arima
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
import os

# 📂 Create a folder for saving results
output_folder = "Retail_Store_2"
os.makedirs(output_folder, exist_ok=True)

# 📌 Load Dataset
file_path = "../../data/raw/store2.csv"  # Ensure this is the correct path
data = pd.read_csv(file_path, encoding="utf-8")

# ✅ Convert 'SaleDate' to datetime format
data['SaleDate'] = pd.to_datetime(data['SaleDate'], format='%d/%m/%Y')

# 🔹 Get list of all unique products
products = data['ProductName'].unique()

# 🔄 Loop through all products
for selected_product in products:
    print(f"Processing: {selected_product}...")

    # Extract & process data for the selected product
    product_data = data[data['ProductName'] == selected_product]

    # 📊 Aggregate sales by month
    monthly_sales = product_data.resample('M', on='SaleDate')['QuantitySold'].sum().fillna(0)

    # 🚨 Ensure enough data for training
    if len(monthly_sales) < 24:
        print(f"Skipping {selected_product} due to insufficient data.")
        continue

    train = monthly_sales["2015-01-01":"2023-12-31"]

    # 🔮 Forecast for 60 months (2024-2028)
    forecast_steps = 60
    auto_model = auto_arima(train, seasonal=True, m=12, d=1, D=1, stepwise=True, trace=True)
    best_order = auto_model.order
    best_seasonal_order = auto_model.seasonal_order

    # 🎯 Train ARIMA model
    arima_model = ARIMA(train, order=best_order, seasonal_order=best_seasonal_order)
    arima_results = arima_model.fit()
    arima_forecast = arima_results.forecast(steps=forecast_steps)

    # 🔄 LSTM Model Training & Prediction
    scaler = StandardScaler()

    # ✅ Ensure residuals length match
    min_steps = min(len(train), forecast_steps)
    residuals = arima_forecast.values[:min_steps] - train[-min_steps:].values
    residuals_scaled = scaler.fit_transform(residuals.reshape(-1, 1))

    # ✅ Function to create LSTM sequences
    def create_lstm_data(series, time_steps=3):
        X, y = [], []
        for i in range(len(series) - time_steps):
            X.append(series[i:i + time_steps])
            y.append(series[i + time_steps])
        return np.array(X), np.array(y)

    time_steps = 3
    X_train, y_train = create_lstm_data(residuals_scaled, time_steps)

    # 🚨 Skip if not enough LSTM training samples
    if len(X_train) == 0:
        print(f"Skipping {selected_product} due to insufficient LSTM training samples.")
        continue

    X_train = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))

    # 🤖 Define LSTM Model
    lstm_model = Sequential([
        LSTM(100, activation='relu', return_sequences=True, input_shape=(time_steps, 1)),
        Dropout(0.2),
        LSTM(50, activation='relu'),
        Dropout(0.2),
        Dense(1)
    ])
    lstm_model.compile(optimizer='adam', loss='mse')
    lstm_model.fit(X_train, y_train, epochs=100, batch_size=16, verbose=0)

    # 📈 Make LSTM Predictions
    forecast_steps = min(forecast_steps, len(X_train))
    lstm_pred = lstm_model.predict(X_train[-forecast_steps:].reshape((forecast_steps, time_steps, 1))).flatten()
    lstm_pred = scaler.inverse_transform(lstm_pred.reshape(-1, 1)).flatten()
    final_forecast = 0.95 * arima_forecast[:forecast_steps] + 0.05 * lstm_pred
    final_forecast = np.maximum(1, np.round(final_forecast)).astype(int)

    # 🗓️ Create Forecast Dates
    forecast_dates = pd.date_range(start=monthly_sales.index[-1], periods=forecast_steps + 1, freq='M')[1:]

    # ✅ Save forecasted sales
    forecast_file = f"{output_folder}/{selected_product}_forecast_2024_2028.csv"
    pd.DataFrame({'Date': forecast_dates, 'Predicted Sales': final_forecast}).to_csv(forecast_file, index=False)

    print(f"✅ Forecasting completed for {selected_product}!")
    print(f"📂 Results saved in: {forecast_file}")

    # 📊 **Plot Historical & Forecasted Sales**
    plt.figure(figsize=(14, 7))
    plt.plot(monthly_sales.index, monthly_sales.values, label="Actual Sales (2015-2023)", marker="o", color="blue")
    plt.plot(forecast_dates, final_forecast, label="Predicted Sales (2024-2028)", marker="o", linestyle="--", color="red")

    # Ensure continuity between actual sales and forecast
    plt.plot([monthly_sales.index[-1], forecast_dates[0]], [monthly_sales.values[-1], final_forecast[0]], linestyle="--", color="red")

    plt.title(f"Total Sales + Forecast for {selected_product}")
    plt.xlabel("Date")
    plt.ylabel("Quantity Sold")
    plt.legend()
    plt.grid(True)

    # ✅ Save the updated plot
    plot_file_combined = f"{output_folder}/{selected_product}_forecast_plot.png"
    plt.savefig(plot_file_combined, dpi=300, bbox_inches="tight")

    print(f"📊 Combined sales plot saved as: {plot_file_combined}")

    # ✅ Show the updated plot
    plt.show()
```

```python
import pandas as pd
import os

# 📂 Define folders
actual_sales_file = "R2.csv"
forecast_folder = "Retail_Store_2"
output_folder = "Retail_Store_2Merged"

# ✅ Create output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

# ✅ Load actual sales data
data = pd.read_csv(actual_sales_file)
data['SaleDate'] = pd.to_datetime(data['SaleDate'], format='%d/%m/%Y')

# ✅ Aggregate actual sales by month
actual_sales = data.groupby(['ProductName', pd.Grouper(key='SaleDate', freq='M')])['QuantitySold'].sum().reset_index()

# 🔹 Get list of all unique products
products = data['ProductName'].unique()

# 🔄 Loop through all products and merge actual & predicted sales
for selected_product in products:
    print(f"Processing: {selected_product}...")

    # ✅ Get actual sales for the product
    product_sales = actual_sales[actual_sales['ProductName'] == selected_product]

    # ✅ Try to load the forecasted sales CSV (Skip if not found)
    forecast_file = os.path.join(forecast_folder, f"{selected_product}_forecast_2024_2028.csv")
    if not os.path.exists(forecast_file):
        print(f"❌ No forecast data found for {selected_product}. Skipping.")
        continue
    
    forecast_data = pd.read_csv(forecast_file)
    forecast_data['Date'] = pd.to_datetime(forecast_data['Date'])

    # ✅ Merge actual and forecasted sales
    combined_sales = pd.concat([
        product_sales.rename(columns={'SaleDate': 'Date', 'QuantitySold': 'Sales'}),
        forecast_data.rename(columns={'Predicted Sales': 'Sales'})
    ])

    # ✅ Save merged sales data
    output_file = os.path.join(output_folder, f"{selected_product}_total_and_forecasted_sales.csv")
    combined_sales.to_csv(output_file, index=False)

    print(f"✅ Saved merged sales data for {selected_product}: {output_file}")
```

```python
import pandas as pd
import os

# 📂 Define the folder containing merged sales data
merged_folder = "Retail_Store_2Merged"
output_file = "ALLRetail_Store_2.xlsx"

# ✅ Get list of all merged sales files
merged_files = [f for f in os.listdir(merged_folder) if f.endswith("_total_and_forecasted_sales.csv")]

# ✅ Read and combine all merged sales data
all_sales_data = []

for file in merged_files:
    product_name = file.replace("_total_and_forecasted_sales.csv", "")  # Extract product name
    file_path = os.path.join(merged_folder, file)
    
    # Load sales data
    df = pd.read_csv(file_path)
    df['Date'] = pd.to_datetime(df['Date'])
    df['ProductName'] = product_name  # Add product name column
    
    all_sales_data.append(df)

# ✅ Combine all products into a single DataFrame
merged_sales_df = pd.concat(all_sales_data, ignore_index=True)

# ✅ Save all product sales in **one Excel file**
merged_sales_df.to_excel(output_file, index=False, sheet_name="Sales_Data")

print(f"✅ All product sales saved in: {output_file}")

# ✅ Display first few rows to verify
print(merged_sales_df.head())
```
